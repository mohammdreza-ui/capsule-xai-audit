# capsule-xai-audit on Kaggle

Runs the whole pipeline: preflight, data preparation, smoke test, every experiment, report.

**Before running**
1. Settings → Accelerator: **GPU T4 x2**; Internet: **On** (needs a phone-verified account).
2. Add Input: the **Kvasir-Capsule** labelled images
   ([Simula](https://datasets.simula.no/kvasir-capsule/)) and **Capsule Vision 2024** train/val
   ([figshare](https://figshare.com/articles/dataset/Training_and_Validation_Dataset_of_Capsule_Vision_2024_Challenge/26403469)).
   Any layout works: zip, per-class `.tar.gz`, or extracted folders.
3. Optional: attach the code as a Dataset instead of cloning it from GitHub.
4. **Save Version → Save & Run All** so the run finishes without the browser open.

The full profile takes about 5 h on T4 x2. A stopped run continues when its output is
attached to a new notebook and this notebook is run again.

## 1. Get the code

```python
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/mohammdreza-ui/capsule-xai-audit.git"
CODE = Path("/kaggle/working/capsule-xai-audit")

if not CODE.exists():
    # Prefer a copy attached as a Kaggle Dataset; otherwise clone from GitHub.
    found = subprocess.run(
        "find -L /kaggle/input -maxdepth 5 -name pyproject.toml -path '*capsule-xai-audit*' -print -quit",
        shell=True, capture_output=True, text=True,
    ).stdout.strip()
    if found:
        subprocess.run(["cp", "-r", str(Path(found).parent), str(CODE)], check=True)
        print("code copied from", Path(found).parent)
    else:
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(CODE)], check=True)
        print("code cloned from", REPO_URL)
print(sorted(p.name for p in CODE.iterdir()))
```

## 2. Run

`PROFILE`: `minimum` (Q1 + Q2, ResNet-50, 3 seeds), `core` (Q1–Q4) or `full` (everything).
Extra flags go to `scripts/run_all.py`, e.g. `--seeds 0 1 2` or `--dry-run`.
